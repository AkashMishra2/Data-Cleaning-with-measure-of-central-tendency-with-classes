# Data Cleaning

## Measure of Central Tendency by class

```python
# importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# importing dataset
dataset_org = pd.read_csv("D:\\dataset\\train.csv")
dataset_org.shape
```

```python
# Adjusting rows and columns so that we can see
pd.set_option("display.max_columns",None)
pd.set_option("display.max_rows",None)
```

```python
dataset_org.head(38)
```

```python
# finding missing data 
dataset_org.isnull().sum()
```

```python
# Visualizing the dataset with heatmap
plt.figure(figsize=(8,8))
sns.heatmap(data = dataset_org.isnull())
plt.show()
```

```python
# finding Percentage for every columns 
data_20_Percent = dataset_org.isnull().sum()/dataset_org.shape[0]*100
data_20_Percent
```

```python
# droping columns where more than 20% of data is missing
list_of_20_Percent = data_20_Percent[data_20_Percent>20].keys()
deleted_column_dataset = dataset_org.drop(columns=list_of_20_Percent)
print(deleted_column_dataset.shape)  # shape 
deleted_column_dataset.head(5)
```

```python
# Visualizing the dataset with heatmap
plt.figure(figsize=(8,8))
sns.heatmap(data = deleted_column_dataset.isnull())
plt.show()
```

```python
# We will impute the data in numerical variable.
numeric_dataset = deleted_column_dataset.select_dtypes(include=["int64","float64"])
print(numeric_dataset.shape)
```

```python
# checking that how many columns are of missing values
filtered_numeric_data = numeric_dataset.isnull().sum()
```

```python
# getting missing values columns name in list
missing_INTvalues_columns =list(filtered_numeric_data[filtered_numeric_data!=0].keys())
missing_INTvalues_columns
```

```python
# getting only isnull values from columns
numeric_dataset[missing_INTvalues_columns][numeric_dataset[missing_INTvalues_columns].isnull().any(axis=1)]
```

## We must have knowledge of categorical value of data set to identifying the classes.

```python
# by taking reference of "LotConfig" we will fill "LotFrontage"
# "LotConfig" is catogrical variable

# "LotConfig" before builting logic we have to identify what are the classes

print(dataset_org["LotConfig"].isnull().sum()) # checking the null values in LotConfig variable

dataset_org["LotConfig"].unique()   # all the classes in "LotConfig"
```

```python
# logic
#  .loc[row :,columns]
sliced_dataFrame = dataset_org[dataset_org.loc[:,"LotConfig"] =="Inside"]["LotFrontage"]
sliced_dataFrame.replace(np.nan,sliced_dataFrame.mean())

print(sliced_dataFrame)  # return the values having only "Inside" Class
# But this only for class inside
```

## Filled value with .mean()

```python
df_copy = dataset_org.copy()
for var_class in dataset_org["LotConfig"].unique():
    df_copy.update(dataset_org[dataset_org.loc[:,"LotConfig"] == var_class]["LotFrontage"].replace(
        np.nan , dataset_org[dataset_org.loc[:,"LotConfig"] == var_class]["LotFrontage"].mean()))

# .update() will update the piece of code real time
```

```python
# reference variables must be not have NaN values 

df2_copy = dataset_org.copy()
missing_INTvalues_columns  #  ['LotFrontage', 'MasVnrArea', 'GarageYrBlt']

# reference_class_var = ["LotConfig","MasVnrType","GarageType"]  
#  Are not cleaned or Having NaN values

reference_class_var = ["LotConfig","Exterior2nd","KitchenQual"]

for reference_var , missing_NUM_columns in zip(reference_class_var,missing_INTvalues_columns):
    for var_class in dataset_org[reference_var].unique():
        df2_copy.update(dataset_org[dataset_org.loc[:,reference_var]==var_class][missing_NUM_columns].replace(
            np.nan,dataset_org[dataset_org.loc[:,reference_var]==var_class][missing_NUM_columns].mean()
        ))
```

```python
df2_copy.select_dtypes(include=["int64","float64"]).isnull().sum()
```

# Data Distribution for Mean

```python
sns.set()  # for grid
for i,var in enumerate(missing_INTvalues_columns):
    plt.figure(figsize=(15,15))
    plt.subplot(2,2,i+1)
    sns.distplot(dataset_org[var],kde_kws={"linewidth":5,"color":"red"},label="original")
    sns.distplot(df2_copy[var],kde_kws={"linewidth":3,"color":"green"},label="cleaned")
    plt.legend(loc=1)
    plt.show()
```

# Handling missing value with Median()

```python
# reference variables must be not have NaN values 

df3_copy = dataset_org.copy()
missing_INTvalues_columns  #  ['LotFrontage', 'MasVnrArea', 'GarageYrBlt']

# reference_class_var = ["LotConfig","MasVnrType","GarageType"]  
#  Are not cleaned or Having NaN values

reference_class_var = ["LotConfig","Exterior2nd","KitchenQual"]

for reference_var , missing_NUM_columns in zip(reference_class_var,missing_INTvalues_columns):
    for var_class in dataset_org[reference_var].unique():
        df3_copy.update(dataset_org[dataset_org.loc[:,reference_var]==var_class][missing_NUM_columns].replace(
            np.nan,dataset_org[dataset_org.loc[:,reference_var]==var_class][missing_NUM_columns].median()
        ))
```

```python
df3_copy.select_dtypes(include=["int64","float64"]).isnull().sum()
```

# Data Distribution for Median

```python
sns.set()  # for grid
for i,var in enumerate(missing_INTvalues_columns):
    plt.figure(figsize=(15,15))
    plt.subplot(2,2,i+1)
    sns.distplot(dataset_org[var],kde_kws={"linewidth":5,"color":"red"},label="original")
    sns.distplot(df3_copy[var],kde_kws={"linewidth":3,"color":"green"},label="cleaned")
    plt.legend(loc=1)
    plt.show()
```

# Checking Distribution of original dataset with mean and median

```python
sns.set()  # for grid
for i,var in enumerate(missing_INTvalues_columns):
    plt.figure(figsize=(15,15))
    plt.subplot(2,2,i+1)
    sns.distplot(dataset_org[var],kde_kws={"linewidth":6,"color":"red"},label="original")
    sns.distplot(df2_copy[var],kde_kws={"linewidth":4,"color":"green"},label="mean")
    sns.distplot(df3_copy[var],kde_kws={"linewidth":2,"color":"yellow"},label="median")
    plt.legend(loc=1)
    plt.show()
```

# Checking outliers with the help of boxplot

```python
# Checking Outliers
for i,var in enumerate(missing_INTvalues_columns):
    plt.figure(figsize=(10,10))
    plt.subplot(3,1,1)
    sns.boxplot(dataset_org[var],orient="h")
    plt.subplot(3,1,2)
    sns.boxplot(df2_copy[var],orient="h")
    plt.subplot(3,1,3)
    sns.boxplot(df3_copy[var],orient="h")    
```
